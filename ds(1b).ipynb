```python
import pandas as pd

student_data = {
    "Roll_No": [101, 102, 103, 104],
    "Name": ["priya", "divya", "hema", "preethi"],
    "department": ["csd", "cse", "cse", "csd"],
    "percentage": [98, 99, 99, 98]
}

df = pd.DataFrame(student_data)

df.to_json("student.json", orient="records", indent=4)

print("JSON file created successfully")
print(df.to_json(orient="records", indent=4))
    
```

```python
import pandas as pd

df = pd.read_json("student.json")   

print("student_data")
print(df)
```

```python
import pandas as pd
from io import StringIO

json_data = """
[
    {
        "Roll_No": 101,
        "Name": "priya",
        "department": "csd",
        "percentage": 98
    },
    {
        "Roll_No": 102,
        "Name": "divya",
        "department": "cse",
        "percentage": 99
    },
    {
        "Roll_No": 103,
        "Name": "hema",
        "department": "cse",
        "percentage": 99
    }
]
"""

df = pd.read_json(StringIO(json_data))

print("Parsed JSON data")
print(df)
```
