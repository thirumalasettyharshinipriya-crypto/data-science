```python
import requests
import pandas as pd

# GitHub API URL
url = "https://api.github.com/repos/pandas-dev/pandas/issues"

# Send GET request
response = requests.get(url)

# Convert JSON response
issues = response.json()

# Create DataFrame
df_issues = pd.DataFrame(issues)

# Display selected columns
print(df_issues[["id", "title", "state", "created_at"]].head())
```

```python
# SQL Database Operations using SQLite

import sqlite3
import pandas as pd

# Connect to database
conn = sqlite3.connect("students.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER,
    name TEXT
)
""")

# Insert records
cursor.execute("INSERT INTO students VALUES (1, 'Harshini')")
cursor.execute("INSERT INTO students VALUES (2, 'Preethi')")

conn.commit()

# Read data
df = pd.read_sql_query("SELECT * FROM students", conn)
print("Student Records:")
print(df)

# Update record
cursor.execute("""
UPDATE students
SET name = 'pandu'
WHERE id = 1
""")

conn.commit()

print("\nAfter Update:")
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)

# Delete record
cursor.execute("""
DELETE FROM students
WHERE id = 1
""")

conn.commit()

print("\nAfter Delete:")
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)

# Close connection
conn.close()
```
