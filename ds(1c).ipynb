```python
import pandas as pd
student_data = {
    "Roll_no": [101, 102, 103],
    "Name": ["pandu", "harshini", "priya"],
    "Department": ["DS", "CS", "IT"],
    "Percentage": [98, 99, 99]
}
course_data = {
    "course_ID": ["c101", "c102", "c103"],
    "course_name": ["Python", "Data science", "Machine learning"],
    "Credits": [4, 4, 3]
}
student_df = pd.DataFrame(student_data)
course_df = pd.DataFrame(course_data)
with pd.ExcelWriter("college_data.xlsx", engine="openpyxl") as writer:
    student_df.to_excel(
        writer,
        sheet_name="students",
        index=False
    )
    course_df.to_excel(
        writer,
        sheet_name="courses",
        index=False
    )
print("Multiple sheets successfully written to college_data.xlsx")
```

```python
import pandas as pd
import json
from io import StringIO

# 1. CSV READING AND WRITING

csv_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

csv_df = pd.read_csv(csv_url)

print("\nCSV Data:")
print(csv_df.head())

# Writing CSV file
csv_df.to_csv("iris-output.csv", index=False)

print("\nCSV file written successfully.")

# 2. TAB-DELIMITED FILE READING AND WRITING

student_data = {
    "Roll-No": [101, 102, 103, 104],
    "Name": ["Priya", "Preethi", "Vinay", "Lohith"],
    "Department": ["CSD", "CSE", "CSE", "CSD"],
    "Percentage": [98, 99, 99, 98]
}

student_df = pd.DataFrame(student_data)

# Writing tab-delimited file
student_df.to_csv(
    "student-output.txt",
    sep="\t",
    index=False
)

print("\nTab-delimited file written successfully.")

# Reading tab-delimited file
tab_df = pd.read_csv(
    "student-output.txt",
    sep="\t"
)

print("\nTab-Delimited Data:")
print(tab_df)
# 3. JSON READING AND WRITING
student_df.to_json(
    "student-output.json",
    orient="records",
    indent=4
)
# Reading JSON file
json_df = pd.read_json("student-output.json")

print("\nJSON Data:")
print(json_df)
# Parsing JSON String
json_string = """
[
    {
        "Roll-No": 201,
        "Name": "Priya",
        "Percentage": 92
    },
    {
        "Roll-No": 202,
        "Name": "Preethi",
        "Percentage": 87
    }
]
"""

# Convert JSON string into Python object
parsed_json = json.loads(json_string)

print("\nParsed JSON String:")
print(parsed_json)

# Convert parsed JSON into DataFrame
parsed_json_df = pd.DataFrame(parsed_json)

print("\nParsed JSON DataFrame:")
print(parsed_json_df)

# 4. EXCEL READING AND WRITING
# Write DataFrame to Excel
student_df.to_excel(
    "students.xlsx",
    sheet_name="Student Details",
    index=False
)

# Read Excel file
excel_df = pd.read_excel(
    "students.xlsx",
    sheet_name="Student Details"
)

print("\nExcel Data:")
print(excel_df)

print("\nAll file operations completed successfully.")
```
