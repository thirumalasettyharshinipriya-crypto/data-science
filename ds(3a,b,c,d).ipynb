```python
import numpy as np
import pandas as pd
data = {
    'A': [1, 2, np.nan],
    'B': [4, np.nan, np.nan]
}
df = pd.DataFrame(data)
print(df.isna())
df_clean = df.dropna()
df_filled = df.fillna(0)
print("\nDataFrame after dropna:")
print(df_clean)
print("\nDataFrame after fillna:")
print(df_filled)
```

```python
df=pd.DataFrame({'gender':['male','female','male']})
df['gender_name']=df['gender'].map({'male':0,'female':1})
df_renamed=df.rename(columns={'gender':'Gender'})
print(df_renamed)
```
